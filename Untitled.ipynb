```python
import cv2
import numpy as np

def preprocess_for_ocr(image_path, output_path=None):
    # 1. อ่านไฟล์ภาพ
    image = cv2.imread(image_path)
    
    # 2. แปลงเป็นภาพเกรย์สเกล (Grayscale)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # 3. เพิ่มความคมชัดและปรับแสงไม่เท่ากันด้วย CLAHE
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    enhanced = clahe.apply(gray)
    
    # 4. ลดสัญญาณรบกวนและจุดเปื้อน (Noise Reduction)
    denoised = cv2.medianBlur(enhanced, 3)
    
    # 5. แปลงเป็นภาพขาว-ดำบริสุทธิ์ด้วย Otsu's Thresholding
    _, binary = cv2.threshold(denoised, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # บันทึกไฟล์ที่ประมวลผลแล้ว
    if output_path:
        cv2.imwrite(output_path, binary)
        
    return binary

# ตัวอย่างการใช้งาน
processed_img = preprocess_for_ocr('example1.jpg', 'ocr_ready.png')
```

```python
import easyocr

# กำหนด gpu=False เพื่อบังคับให้รันบน CPU
reader = easyocr.Reader(['th'], gpu=False)

# ประมวลผลภาพ
results = reader.readtext('ocr_ready.png')

for result in results:
    text = result[1]
    confidence = result[2]
    print(f"ข้อความ: {text} (ความแม่นยำ: {confidence:.2f})")
```
